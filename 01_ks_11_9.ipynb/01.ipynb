### CF and KNN, Neighborhood Formation

#### Importing the Libraries

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity as sim
```

#### Load the Dataset

```python
df = pd.read_csv('/code/github/Kapita-Selekta/01_ks_11_9.ipynb/cf_data_matrix.csv')

df.head()
```

#### Rate and Fill the Nan Value

```python
rating_matrix = df.pivot(index='user', columns="item", values="rating").fillna(0)

rating_matrix
```

#### Compute

```python
user_sim = sim(rating_matrix)

user_sim
```

```python
df_user_sim = pd.DataFrame(user_sim, index=rating_matrix.index, columns=rating_matrix.index)

df_user_sim
```

```python
kandidate = df_user_sim['U1'].sort_values(ascending=False)

kandidate
```

```python
kandidate.head(7)
```

```python
# nearest = 7
k_nb = kandidate.index.to_list()[1:8]

k_nb
```

#### Item-Based

```python
item_sim = sim(rating_matrix.T)

item_sim
```

```python
df_item_sim = pd.DataFrame(item_sim, index = rating_matrix.T.index,columns = rating_matrix.T.index)

df_item_sim
```

```python
kandidate = df_item_sim['I1'].sort_values(ascending=False)

kandidate
```

```python
# nearest = 7
k_nb = kandidate.index.to_list()[1:8]

k_nb
```
